```python
import torch
import torch.nn.functional as F
```

```python
x = torch.tensor([-2., -1., 0., 1., 2.])
```

```python
def swish(x):
    return x * torch.sigmoid(x)
```

```python
y = swish(x)

print("Input:", x)
print("Swish:", y)
```

```python
pytorch_y = F.silu(x)

print("PyTorch:", pytorch_y)
```

```python
print(torch.allclose(y, pytorch_y))
```

```python
import matplotlib.pyplot as plt

x = torch.linspace(-5, 5, 100)

y = swish(x)

plt.plot(x, y)
plt.title("Swish Activation Function")
plt.xlabel("x")
plt.ylabel("Swish(x)")
plt.grid()
plt.show()
```
